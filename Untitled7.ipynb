```python
import numpy as np

X = np.array([
    [1, 2, 1],
    [0, 1, 0],
    [2, 1, 2]
])


W_q = np.random.randn(3, 3)
W_k = np.random.randn(3, 3)
W_v = np.random.randn(3, 3)


Q = np.dot(X, W_q)
K = np.dot(X, W_k)
V = np.dot(X, W_v)


scores = np.matmul(Q, K.T)

scale = np.sqrt(K.shape[1])
scores = scores / scale

exp_scores = np.exp(scores)
attention = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)


result = np.matmul(attention, V)

print("Attention Matrix:\n", attention)
print("Final Output:\n", result)
```
